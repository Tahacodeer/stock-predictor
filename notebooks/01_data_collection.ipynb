```python
import yfinance as yf
import datetime as dt
import pandas as pd
import os
def data_collection(ticker,hdd):
    end_date=dt.date.today()
    start_date=end_date-dt.timedelta(days=hdd)
    data_rang=yf.download(ticker, start=start_date, end=end_date)
    data_rang = pd.DataFrame(data_rang)
    os.makedirs("/Users/tahamac/stock-predictor/data/raw", exist_ok=True)
    filename = f"/Users/tahamac/stock-predictor/data/raw/{ticker}_{start_date}_{end_date}.csv"
    data_rang.to_csv(filename, index=False)
    return 

ticker=input("entre the ticker ")
hdd=int(input("entre the days range "))
data_collection(ticker,hdd)

 
```
